```python
import pandas as pd
import matplotlib.pyplot as plt

# Đọc CSV không có header
df = pd.read_csv(
    "data_dht.csv",
    header=None,
    names=["timestamp", "temperature", "humidity", "valid"]
)

print(df.head())

# Tính MA5
df["MA5"] = df["temperature"].rolling(window=5).mean()

# Vẽ đồ thị
plt.figure(figsize=(12,5))

plt.plot(
    df["timestamp"],
    df["temperature"],
    label="Temperature Raw",
    marker="o"
)

plt.plot(
    df["timestamp"],
    df["MA5"],
    label="Temperature MA5",
    linewidth=3
)

plt.xlabel("Time (ms)")
plt.ylabel("Temperature (°C)")
plt.title("DHT11 Temperature Filtering")

plt.legend()
plt.grid(True)

plt.show()

plt.figure(figsize=(12,5))

plt.plot(df["timestamp"],
         df["humidity"],
         marker="o")

plt.xlabel("Time (ms)")
plt.ylabel("Humidity (%)")
plt.title("DHT11 Humidity")
plt.grid(True)

plt.show()
```
